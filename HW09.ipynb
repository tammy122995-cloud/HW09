### 兩場活動報名整理器：核心概念說明

#### 1. 集合（Set）運算及其意義
在 Python 中，`set` 是一種「不重複、無順序」的容器，非常適合用來進行名單的比對：
* **交集 (Intersection `&`)**：兩場活動都有參加的人。代表同時出現在桌遊與電影名單中的共同名單。
* **聯集 (Union `|`)**：至少參加了一場活動的人。將兩場活動的所有參加者合併，且重疊的人只會出現一次。
* **差集 (Difference `-`)**：
  * **桌遊 - 電影 (雙向差集之一)**：只參加桌遊、而沒有參加電影欣賞的人。
  * **電影 - 桌遊 (雙向差集之二)**：只參加電影欣賞、而沒有參加桌遊體驗的人。

#### 2. 「人數」與「人次」的差別
* **總不重複人數（人數）**：代表實際上「有多少個物理人」參與了活動。只要小明兩場都參加，他也只會被計算為「1人」（聯集 `len(union)`）。
* **報名人次（人次）**：代表「兩場活動報名表上的獨立名額總和」。如果小明兩場都報名，在桌遊算 1 人次，在電影也算 1 人次，因此共累積「2人次」（桌遊去重人數 + 電影去重人數）。這通常用於評估場地容納量、總工作量等。

```python
# ==========================================
# 兩場活動報名整理器 (測試資料一)
# ==========================================

# 0. 基礎設定資料 (由元組保存)
activity_names = ("桌遊體驗", "電影欣賞")
unit_fees = {"桌遊體驗": 120.0, "電影欣賞": 80.0}

# 測試資料一
board_raw = " Amy , ben, amy, Cora "
movie_raw = "ben, dora, Eric, BEN"

# ------------------------------------------
# 1. 原始資料清理與拆分
# ------------------------------------------
# 先移除所有半形空白，並將英文字母全部轉成小寫
board_clean = board_raw.replace(" ", "").lower()
movie_clean = movie_raw.replace(" ", "").lower()

# 依據逗號拆開成串列 (保留重複、保留原始順序)
board_list = board_clean.split(",")
movie_list = movie_clean.split(",")

# ------------------------------------------
# 2. 集合去重與比較名單
# ------------------------------------------
# 轉為集合自動去重
board_set = set(board_list)
movie_set = set(movie_list)

# 計算各種集合運算
common_participants = board_set & movie_set          # 交集：共同參加
all_participants = board_set | movie_set             # 聯集：全部不重複參加者
only_board = board_set - movie_set                   # 差集：只參加桌遊
only_movie = movie_set - board_set                   # 差集：只參加電影

# 排序集合內容以便輸出 (轉為 sorted list)
sorted_board = sorted(board_set)
sorted_movie = sorted(movie_set)
sorted_common = sorted(common_participants)
sorted_only_board = sorted(only_board)
sorted_only_movie = sorted(only_movie)
sorted_all = sorted(all_participants)

# ------------------------------------------
# 3. 取得通知名單預覽 (切片取得前兩位)
# ------------------------------------------
preview_list = sorted_all[:2]

# ------------------------------------------
# 4. 建立摘要字典並計算費用
# ------------------------------------------
# 自 unit_fees 元組中提取單價 (使用 activity_names 解包或索引)
board_name = activity_names[0]
movie_name = activity_names[1]
board_price = unit_fees[board_name]
movie_price = unit_fees[movie_name]

# 計算各項數值
count_board = len(board_set)
count_movie = len(movie_set)
count_common = len(common_participants)
count_unique_people = len(all_participants)
count_total_registrations = count_board + count_movie

fee_board = count_board * board_price
fee_movie = count_movie * movie_price
fee_total = fee_board + fee_movie

# 將結果打包存入 summary 字典
summary = {
    "board_count": count_board,
    "movie_count": count_movie,
    "common_count": count_common,
    "unique_people": count_unique_people,
    "total_registrations": count_total_registrations,
    "board_fee": fee_board,
    "movie_fee": fee_movie,
    "total_fee": fee_total
}

# ------------------------------------------
# 5. 格式化輸出完整摘要
# ------------------------------------------
print("==================== 活動報名整理摘要 ====================")
print(f"【活動基本資訊】")
print(f"  * 活動一：{board_name} (每人 {board_price:.2f} 元)")
print(f"  * 活動二：{movie_name} (每人 {movie_price:.2f} 元)")
print()
print(f"【1. 原始清理名單 (保留重複項目)】")
print(f"  * 桌遊體驗原始：{board_list}")
print(f"  * 電影欣賞原始：{movie_list}")
print()
print(f"【2. 兩場各自去重與排序】")
print(f"  * 桌遊去重名單：{sorted_board} (共 {summary['board_count']} 人)")
print(f"  * 電影去重名單：{sorted_movie} (共 {summary['movie_count']} 人)")
print()
print(f"【3. 交叉比對與名單分析】")
print(f"  * 共同參加者 (交集)：{sorted_common}")
print(f"  * 只參加桌遊者 (差集)：{sorted_only_board}")
print(f"  * 只參加電影者 (差集)：{sorted_only_movie}")
print()
print(f"【4. 總體參與統計】")
print(f"  * 全部參加者 (聯集)：{sorted_all}")
print(f"  * 總不重複人數 (實際物理人數)：{summary['unique_people']} 人")
print(f"  * 總報名人次 (活動總人次)：{summary['total_registrations']} 人次")
print(f"  * 前兩位通知預覽：{preview_list}")
print()
print(f"【5. 收費財務摘要】")
print(f"  * 桌遊預計應收：{summary['board_fee']:.2f} 元")
print(f"  * 電影預計應收：{summary['movie_fee']:.2f} 元")
print(f"  * 總計預計應收：{summary['total_fee']:.2f} 元")
print("==========================================================")
```
